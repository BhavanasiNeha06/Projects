```python
import pandas as pd
file_path = "../data/raw/Water_Consumption_Data_for_year.csv"
df = pd.read_csv(file_path, sep="\t")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
```

```python
print(df.columns.tolist())
```

```python

df.columns = [
    "month",
    "ward_number",
    "ward_name",
    "connections",
    "consumption_ml"
]

print(df.columns.tolist())
```

```python
df.info()
```

```python
# Convert month from text to datetime

df["month"] = pd.to_datetime(
    df["month"],
    format="%d-%m-%Y"
)

print(df.dtypes)
```

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Date range:", df["month"].min(), "to", df["month"].max())
print("Unique months:", df["month"].nunique())
```

```python
df["ward_name"] = (
    df["ward_name"]
    .str.strip()
    .str.title()
)
```

```python
df["location_id"] = (
    df["ward_number"].astype(str)
    + "_"
    + df["ward_name"].str.lower().str.replace(" ", "_")
)
```

```python
print(
    "Duplicate Location-Month:",
    df.duplicated(
        subset=["location_id", "month"]
    ).sum()
)
```

```python
print(df.isnull().sum())
```

```python
print("Negative connections:",
      (df["connections"] < 0).sum())

print("Zero connections:",
      (df["connections"] == 0).sum())

print("Negative consumption:",
      (df["consumption_ml"] < 0).sum())

print("Zero consumption:",
      (df["consumption_ml"] == 0).sum())
```

```python
df.loc[
    df["consumption_ml"] < 0,
    "consumption_ml"
] = 0
```

```python
output_file = "../data/processed/bengaluru_water_consumption_cleaned.csv"

df.to_csv(
    output_file,
    index=False
)

print("Saved:", output_file)
```
